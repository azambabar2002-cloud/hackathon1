```python
import pandas as pd

# ---- 1. Load (upload the CSV to Colab first, or change the path) ----
PATH = "hackathon_preli_dataset - bangladesh_dengue_large_dataset.csv"
df = pd.read_csv(PATH, parse_dates=["date"])

# ---- 2. Drop insignificant / redundant columns ----
df = df.drop(columns=["max_temp_c", "min_temp_c"])

# ---- 3. Sort and validate: weekly rows, no gaps, per ward ----
df = df.drop_duplicates(["ward_id", "date"]).sort_values(["ward_id", "date"])
gaps = df.groupby("ward_id")["date"].diff().dropna()
assert (gaps == pd.Timedelta(days=7)).all(), "Dates are not strictly weekly per ward"

# ---- 4. Lag features, computed separately per ward ----
g = df.groupby("ward_id")
lags = {f"rainfall_mm_lag{k}w": g["rainfall_mm"].shift(k) for k in range(1, 5)}
lags["avg_temp_c_lag1w"] = g["avg_temp_c"].shift(1)
lags["relative_humidity_pct_lag1w"] = g["relative_humidity_pct"].shift(1)
df = pd.concat([df, pd.DataFrame(lags)], axis=1)
LAG_COLS = list(lags)

# ---- 5. Target: total dengue cases in the next 4 weeks (t+1 ... t+4), per ward ----
g = df.groupby("ward_id")["dengue_cases"]
df["cases_next4w"] = sum(g.shift(-k) for k in range(1, 5))

# Drop rows with undefined lags (first 4 weeks) or no future data (last 4 weeks)
df = df.dropna(subset=LAG_COLS + ["cases_next4w"]).reset_index(drop=True)

# ---- 6. Four risk classes: Low <=1, Medium 2-19, High 20-37, Critical >=38 ----
df["risk_class"] = pd.cut(
    df["cases_next4w"],
    bins=[-float("inf"), 1, 19, 37, float("inf")],
    labels=["Low", "Medium", "High", "Critical"],
)

# ---- 7. Drop identifiers and set final column order ----
STATIC = ["population_density_sqkm", "water_logging_risk_index", "sanitation_score"]
CURRENT = ["rainfall_mm", "avg_temp_c", "relative_humidity_pct"]
df = df[STATIC + CURRENT + LAG_COLS + ["dengue_cases", "cases_next4w", "risk_class"]]

print(df.shape, "| nulls:", df.isna().sum().sum())
print(df["risk_class"].value_counts(normalize=True).round(3))
df.to_csv("dengue_classification.csv", index=False)
df.head(20)
```

```python

train = df[df["date"] < "2024-01-01"]
test  = df[df["date"] >= "2024-01-01"]
X_train, y_train = train.drop(columns=["date", "risk_class"]), train["risk_class"]
X_test,  y_test  = test.drop(columns=["date", "risk_class"]),  test["risk_class"]
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import (accuracy_score, f1_score, classification_report,confusion_matrix)
import pandas as pd, joblib

LABELS = ["Low", "Medium", "High", "Critical"]

# ---- Train ----
model = HistGradientBoostingClassifier(
    max_iter=300, learning_rate=0.05, class_weight="balanced", random_state=42
)
model.fit(X_train, y_train.astype(str))

# ---- Evaluate ----
pred = model.predict(X_test)
```

```python
model
```

```python
print("Accuracy:", round(accuracy_score(y_test.astype(str), pred), 3))
print("Macro F1:", round(f1_score(y_test.astype(str), pred, average="macro"), 3))
print(classification_report(y_test.astype(str), pred, labels=LABELS, digits=3))
print(pd.DataFrame(confusion_matrix(y_test.astype(str), pred, labels=LABELS),
                   index=[f"true {l}" for l in LABELS],
                   columns=[f"pred {l}" for l in LABELS]))
```
